### Contexto

El entorno **FrozenLake-v1** representa un lago congelado donde el agente debe desplazarse desde el punto de inicio (**S**) hasta la meta (**G**) sin caer en los agujeros (**H**).
El lago tiene zonas seguras (**F**) y resbaladizas: las acciones pueden no resultar en el movimiento esperado.

El entorno puede configurarse como:

* **Determinista:** los movimientos siempre resultan como se eligen.
* **Estocástico (slippery=True):** hay probabilidad de resbalar (más desafiante).

El agente recibe:

* **+1** si llega a la meta
* **0** en cualquier otro caso (caminar o caer)
